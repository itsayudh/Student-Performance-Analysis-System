```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/processed/student_performance_datasetv2.csv')
df.head()
```

```python
# Define the boundaries and integer labels
attendance_bins = [65, 75, 83, 91, 100]
attendance_labels = [0, 1, 2, 3]

# Recalculate the attendance_tier with the updated bins
df['attendance_tier'] = pd.cut(
    df['attendance_percentage'], 
    bins=attendance_bins, 
    labels=attendance_labels, 
    include_lowest=True
).astype(int)

# Print the updated value counts sorted by tier index
print("--- Adjusted Attendance Tier Distribution ---")
print(df['attendance_tier'].value_counts().sort_index())
```

```python
df.head()
```

```python
gpa_bins = [0, 2.0, 2.5, 3.0, 4.01]
gpa_labels = [0, 1, 2, 3]

# Create the gpa_tier column
df['gpa_tier'] = pd.cut(
    df['historical_gpa'], 
    bins=gpa_bins, 
    labels=gpa_labels, 
    include_lowest=True
).astype(int)

# Print the value counts sorted by tier index
print("--- GPA Tier Distribution ---")
print(df['gpa_tier'].value_counts().sort_index())
```

```python
df.head()
```

```python
# Create the continuous assessment average column
df['ca_avg'] = (df['quiz_score_avg'] * 0.5) + (df['assignment_score_avg'] * 0.5)

# Calculate and print summary statistics
ca_mean = df['ca_avg'].mean()
ca_std = df['ca_avg'].std()

print("--- Continuous Assessment Average (ca_avg) Summary ---")
print(f"Mean: {ca_mean:.4f}")
print(f"Std:  {ca_std:.4f}")
```

```python
df.head(2)
```

```python
# Calculate the composite rule-based risk score (0.0 to 1.0 normalized scale)
df['rule_risk_score'] = (
    ((100 - df['attendance_percentage']) * 0.4) + 
    ((100 - df['ca_avg']) * 0.3) + 
    ((100 - df['midterm_score']) * 0.3)
) / 100

# View summary statistics to inspect the distribution of risk scores
print("--- Rule Risk Score Distribution Summary ---")
print(df['rule_risk_score'].describe())
```

```python
df.shape
```

```python
# Calculate midterm_relative using subject_difficulty_score as the grouping proxy
df['midterm_relative'] = (
    df['midterm_score'] - 
    df.groupby('subject_difficulty_score')['midterm_score'].transform('mean')
)

# Calculate and print the summary statistics to verify the operation
m_rel_min = df['midterm_relative'].min()
m_rel_max = df['midterm_relative'].max()
m_rel_mean = df['midterm_relative'].mean()

print("--- Midterm Relative Performance (Proxy Grouping) ---")
print(f"Min:  {m_rel_min:.4f}")
print(f"Max:  {m_rel_max:.4f}")
print(f"Mean: {m_rel_mean:.4f}")
```

```python
#checking the coorelation of newly added features with the final_score
new_features = [
    'attendance_tier', 
    'gpa_tier', 
    'ca_avg', 
    'midterm_relative', 
    'rule_risk_score'
]

# 2. Compute the Pearson correlation matrix for these features plus final_score
corr_matrix = df[new_features + ['final_score']].corr()

# 3. Extract final_score's row/column, drop the self-correlation, and sort descending
new_features_corr = corr_matrix['final_score'].drop('final_score').sort_values(ascending=False)

# 4. Print the sorted correlations
print("--- Correlation of Engineered Features against final_score ---")
for feature, r_val in new_features_corr.items():
    print(f"{feature: <20}: r = {r_val:.4f}")
```
